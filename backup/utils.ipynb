```python
from pathlib import Path
import sqlite3


DATA_DIR = Path("data")
DATA_DIR.mkdir(exist_ok=True)

PATIENT_REPORTS_DB = DATA_DIR / "patient_reports.db"
GENERATED_REPORTS_DB = DATA_DIR / "generated_reports.db"
EXTERNAL_MATERIALS_DB = DATA_DIR / "external_materials.db"


def create_databases():
    with sqlite3.connect(PATIENT_REPORTS_DB) as connection:
        connection.execute("""
            CREATE TABLE IF NOT EXISTS patient_reports (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                patient_id TEXT NOT NULL,
                report_type TEXT,
                report_text TEXT NOT NULL,
                file_name TEXT,
                created_at TEXT NOT NULL DEFAULT CURRENT_TIMESTAMP
            )
        """)

    with sqlite3.connect(GENERATED_REPORTS_DB) as connection:
        connection.execute("""
            CREATE TABLE IF NOT EXISTS generated_reports (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                patient_report_id INTEGER NOT NULL,
                patient_id TEXT NOT NULL,
                generated_text TEXT NOT NULL,
                tone TEXT NOT NULL,
                model_name TEXT,
                created_at TEXT NOT NULL DEFAULT CURRENT_TIMESTAMP
            )
        """)

    with sqlite3.connect(EXTERNAL_MATERIALS_DB) as connection:
        connection.execute("""
            CREATE TABLE IF NOT EXISTS external_materials (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                title TEXT NOT NULL,
                source TEXT,
                material_type TEXT,
                content TEXT NOT NULL,
                created_at TEXT NOT NULL DEFAULT CURRENT_TIMESTAMP
            )
        """)

    return "Three databases are ready."


create_databases()
```
